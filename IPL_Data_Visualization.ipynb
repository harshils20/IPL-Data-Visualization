Q1. which team won the most matches?


```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index,x = match_wins.values,palette='viridis')
plt.title("Matches won by each team")
```

Toss Decision Trends

```python
sns.countplot(x = df['toss_decision'],palette='viridis')
plt.title("Toss decision TRends")
```

Toss Winners vs Match winners

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count*100)/df.shape[0]
print(percentage.round(2))
```

How do teams win? (runs or wickets)

```python
sns.countplot(x = df['won_by'],palette='viridis')
plt.title("Won by")
```

Most Player of the match awards?

```python
count = df["player_of_the_match"].value_counts().head(10)
```

```python
sns.barplot(x = count.values, y = count.index,palette='viridis')
plt.title("Top 10 players of the match awards")
```

Top Scorers

```python
hs = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending = False).head(5)
hs
```

```python
hs.plot(kind='barh', color=sns.color_palette('viridis', len(hs)))
```

10 Best bowling figures

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(5)
top_bowlers
```

```python
top_bowlers.plot(kind='barh', color=sns.color_palette('viridis', len(top_bowlers)))
```
